```python
#•	For Loop: Iterates over a sequence (like a list or string)
for i in range(5):
    print(i)
```

```python
#•	While Loop: Continues until a specified condition is false.
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
# Increment a number by 7 until it exceeds 30
x = 12
while x <= 30:
    x = x + 7
    print(x)
```

```python
# Increments a number by 7 and exit the loop before it exceeds 30

x = 12
while x <= 30:
    if (x+7) > 30:
        break
    x = x + 7
    print(x)
```

```python
#Or change the while condition so that x will not increment beyond 30

x = 12;
while (x+7) <= 30:
    x = x + 7
    print(x)
```
