```python
import numpy as np
import pandas as pd

NUM = np.array([1,2,3,4,5,6,])
print(NUM[NUM>=3])
Num = pd.Series(NUM,index=([1,2,3,4,5,6]))
print(Num)

NUM = np.array([1,2,3,4,5,6,])
print(NUM[NUM%2==0])
```
