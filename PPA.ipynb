```python
#Date: 9th November, 2026
#Predictive Analytics
#1. import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

```

```python
# 2. Load the data (ie import the data)
data = pd.read_csv("student_data.csv")
data.tail()
```

```python
data.isna
```

```python
data.info()
```

```python
prediction = []

for index, student in data.iterrows():
    average_score = (student["assignment_pct"] + student["quiz_pct"])/2
    if student["attendance_pct"] < 70 or average_score < 60:
        prediction.append("Unlikely to Pass")
    else:
        prediction.append("Likely to Pass")
```

```python
data["average_scores"] = (data["assignment_pct"] + data["quiz_pct"])/2
data["prediction"] = prediction
data.tail()
```

```python
# 3. Assign input and output variables
X = data[["quiz_pct"]] # Input Variable
y = data["average_scores"] # Output Variable
```

```python
import matplotlib.pyplot as plt
#plt.figure(figsize = (10,6))
plt.scatter(X, y)
plt.xlabel("Quiz (%)", fontsize = 14, fontweight = "bold")
plt.ylabel("Average Scores (%)", fontsize = 14, fontweight = "bold")
plt.grid(True, axis = "y")
plt.tight_layout()
plt.show()
```

```python
# 4. Define a Model 
model = LinearRegression()
model_fit = model.fit(X, y)

```

```python
predicted_data = model_fit.predict(X)
```

```python
plt.scatter(X, y, label = "Raw Data")
plt.plot(X, predicted_data, color = "red", linewidth = 3, label = "Regression Line")
# plt.xlabel("Quiz (%)", fontsize = 14, fontweight = "bold")
# plt.ylabel("Average Scores (%)", fontsize = 14, fontweight = "bold")
plt.grid(True, axis = "y")
plt.title("Predictive Analytics: Average Scores Dependent on Quiz Scores")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
new_quiz_score = 50
new_data = pd.DataFrame({"quiz_pct":[new_quiz_score]})
new_prediction = model_fit.predict(new_data)[0]
display(f"Estimated Average Score for {new_quiz_score} % is : {new_prediction:.2f} %")
```
