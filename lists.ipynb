```python
#creating and formating list items
#index of an array starts from 0
special_numbers = [11,22,46,77,98,109]
print("list items in special numbers".title())
for num in special_numbers:
    print(num)

car_makers = ['BMW', 'Honda', 'Kia', 'Dosh']
print("list of items in Cars".title())
for car in car_makers:
    print(car)  #prints the list

car_makers[3] = 'Toyota' #replaces Dosh for Toyota in the list
print(car_makers[3])

fruits= []
fruits.append('Apple')
fruits.append('Pawpaw')
fruits.append('Banana')
fruits.append('Orange')
fruits.append('Mango')
fruits.insert(0,'Grape')  #inserting an element in a list using position

print("list items in Fruits".title())
for fruit in fruits:
    print (fruit)

last_fruit = fruits.pop()
print(f"The Last fruit that was added is {last_fruit}")

fruits.sort()
print (fruits)

```
